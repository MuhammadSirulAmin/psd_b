# Statistik Data Kualitas Udara Menggunakan KNIME

## 1. penjelasan

Analisis statistik digunakan untuk memahami karakteristik dan pola suatu data. Pada penelitian ini, analisis dilakukan terhadap dataset kualitas udara wilayah Klampis, Kabupaten Bangkalan menggunakan KNIME Analytics Platform.

Data yang digunakan disimpan dalam database PostgreSQL pada Aiven. Data tersebut kemudian dihubungkan ke KNIME menggunakan PostgreSQL Connector dan DB Query Reader. Setelah data berhasil diambil, analisis statistik dilakukan menggunakan node Statistics.

Dataset memiliki 8.688 baris data dan beberapa variabel kualitas udara, yaitu:

  - PM2.5
  - PM10
  - Carbon Monoxide (CO)
  - Nitrogen Dioxide (NO2)
  - Sulphur Dioxide (SO2)
  - Ozone (O3)

Selain variabel kualitas udara, dataset juga memiliki kolom waktu, wilayah, latitude, dan longitude.

```sql
SELECT *
FROM kualitas_udara_data;

2.3 Statistics

Node Statistics digunakan untuk menghitung berbagai ukuran statistik pada data numerik.

Pada konfigurasi Statistics, opsi Calculate median values diaktifkan sehingga nilai median dapat dihitung.

3. Hasil Statistik KNIME

Berdasarkan hasil Statistics pada KNIME, diperoleh hasil sebagai berikut:

| Variabel               | Minimum | Maximum |    Mean | Median | Std. Deviation |   Variance | Skewness | Kurtosis |
| ---------------------- | ------: | ------: | ------: | -----: | -------------: | ---------: | -------: | -------: |
| PM2.5                  |     0.7 |    72.9 |  16.043 |   13.4 |          9.424 |     88.815 |    1.406 |    2.341 |
| PM10                   |     1.1 |    74.3 |  18.056 |   15.9 |          9.472 |     89.723 |    1.268 |    2.072 |
| Carbon Monoxide (CO)   |      83 |   1,935 | 262.592 |    225 |        150.805 | 22,742.115 |    3.840 |   23.034 |
| Nitrogen Dioxide (NO2) |       0 |    30.1 |   5.450 |    3.6 |          4.886 |     23.871 |    1.184 |    0.977 |
| Sulphur Dioxide (SO2)  |       0 |    45.5 |   7.878 |      5 |          7.356 |     54.110 |    1.270 |    1.285 |
| Ozone (O3)             |      15 |     254 |  89.832 |     84 |         32.187 |  1,036.031 |    1.240 |    2.196 |


# 4. Penjelasan Properties pada KNIME

Node **Statistics** pada KNIME digunakan untuk mengetahui karakteristik statistik dari data. Beberapa properties yang ditampilkan antara lain Column, Minimum, Maximum, Mean, Median, Quantile, Standard Deviation, Skewness, Kurtosis, Missing Values, NaN, Infinity, dan Row Count.

---

## 4.1 Column

**Column** menunjukkan nama kolom atau variabel yang sedang dianalisis.

Contoh:

`pm2_5` menunjukkan bahwa statistik tersebut merupakan statistik untuk variabel PM2.5.

---

## 4.2 Minimum

**Minimum** adalah nilai paling kecil yang terdapat dalam suatu kolom.

Contoh pada dataset:

Nilai minimum PM2.5 = **0.7**.

### Contoh perhitungan

Data:

`10, 15, 20, 25, 30`

Maka:

**Minimum = 10**

---

## 4.3 Maximum

**Maximum** adalah nilai paling besar yang terdapat dalam suatu kolom.

Contoh pada dataset:

Nilai maksimum CO = **1935**.

### Contoh perhitungan

Data:

`10, 15, 20, 25, 30`

Maka:

**Maximum = 30**

---

## 4.4 Mean

**Mean** atau rata-rata adalah jumlah seluruh nilai dibagi dengan jumlah data.

### Rumus

\[
Mean = \frac{\sum x}{n}
\]

Contoh pada dataset:

Mean PM2.5 = **16.043**.

### Contoh perhitungan

Data:

`10, 20, 30`

\[
Mean = \frac{10+20+30}{3}
\]

\[
Mean = 20
\]

---

## 4.5 Median

**Median** adalah nilai tengah setelah data diurutkan.

Contoh pada dataset:

Median PM2.5 = **13.4**.

### Contoh perhitungan

Data:

`2, 4, 7, 9, 12`

Nilai tengah = **7**

Jadi Median = **7**.

---

## 4.6 25% Quantile

**25% Quantile atau Q1** menunjukkan batas sekitar 25% data terbawah.

Artinya, sekitar 25% data berada pada atau di bawah nilai Q1.

---

## 4.7 50% Quantile

**50% Quantile** merupakan nilai tengah data dan sama dengan median.

\[
50\% Quantile = Median
\]

Contoh pada PM2.5:

**50% Quantile = 13.4**

---

## 4.8 75% Quantile

**75% Quantile atau Q3** menunjukkan batas sekitar 75% data.

Artinya, sekitar 75% data berada pada atau di bawah nilai Q3.

---

## 4.9 Standard Deviation

**Standard Deviation** atau simpangan baku menunjukkan seberapa jauh data menyebar dari nilai rata-ratanya.

Semakin besar nilainya, semakin besar penyebaran data.

### Contoh pada dataset

Standard deviation CO = **150.805**.

### Contoh perhitungan

Data:

`8, 10, 12`

Mean = 10

Selisih:

- 8 − 10 = −2
- 10 − 10 = 0
- 12 − 10 = 2

Kuadrat selisih:

- 4
- 0
- 4

Jumlah = 8

Variance:

\[
\frac{8}{3}=2.67
\]

Standard deviation:

\[
\sqrt{2.67}\approx1.63
\]

---

## 4.10 Variance

**Variance** menunjukkan tingkat penyebaran data berdasarkan kuadrat selisih setiap data terhadap mean.

### Rumus

\[
Variance = \frac{\sum(x-\mu)^2}{N}
\]

### Contoh perhitungan

Data:

`8, 10, 12`

Mean = 10

\[
Variance = \frac{(8-10)^2+(10-10)^2+(12-10)^2}{3}
\]

\[
Variance = \frac{4+0+4}{3}
\]

\[
Variance = 2.67
\]

---

## 4.11 Skewness

**Skewness** menunjukkan tingkat kemiringan distribusi data.

- Skewness ≈ 0 → relatif simetris
- Skewness > 0 → miring ke kanan
- Skewness < 0 → miring ke kiri

Contoh pada dataset:

CO memiliki **Skewness = 3.840**.

Artinya distribusi CO sangat miring ke kanan.

---

## 4.12 Kurtosis

**Kurtosis** menggambarkan karakteristik ekor distribusi dan kecenderungan munculnya nilai ekstrem.

Secara umum pada excess kurtosis:

- ≈ 0 → mendekati distribusi normal
- > 0 → ekor relatif lebih berat
- < 0 → ekor relatif lebih ringan

Contoh pada dataset:

CO memiliki **Kurtosis = 23.034**.

Artinya distribusi CO memiliki ekor yang sangat berat dan kecenderungan nilai ekstrem yang tinggi.

---

## 4.13 Overall Sum

**Overall Sum** adalah jumlah seluruh nilai dalam suatu kolom.

### Contoh perhitungan

Data:

`10, 20, 30`

\[
Sum = 10+20+30
\]

\[
Sum = 60
\]

---

## 4.14 No. Missings

**No. Missings** menunjukkan jumlah data yang kosong atau tidak tersedia.

Hasil dataset:

**No. Missings = 0**

Artinya tidak terdapat data kosong pada data numerik yang dianalisis.

---

## 4.15 No. NaNs

**No. NaNs** menunjukkan jumlah nilai **Not a Number**.

Hasil dataset:

**No. NaNs = 0**

Artinya tidak terdapat nilai NaN.

---

## 4.16 No. +∞s

**No. +∞s** menunjukkan jumlah nilai positive infinity (+∞).

Hasil dataset:

**No. +∞s = 0**

Artinya tidak terdapat nilai positive infinity.

---

## 4.17 No. -∞s

**No. -∞s** menunjukkan jumlah nilai negative infinity (-∞).

Hasil dataset:

**No. -∞s = 0**

Artinya tidak terdapat nilai negative infinity.

---

## 4.18 Row Count

**Row Count** menunjukkan jumlah seluruh baris atau observasi dalam dataset.

Hasil dataset:

**Row Count = 8.688**

Artinya terdapat 8.688 observasi.

### Contoh perhitungan

Jika data terdiri dari:

`10, 20, 30, 40, 50`

maka:

**Row Count = 5**

---

## 4.19 Mean Absolute Deviation

**Mean Absolute Deviation (MAD)** menunjukkan rata-rata jarak absolut setiap data terhadap nilai mean.

### Rumus

\[
MAD = \frac{\sum |x-\bar{x}|}{n}
\]

MAD digunakan untuk melihat seberapa jauh data secara rata-rata dari nilai mean.

---

# 5. Hasil Statistik

| Variabel | Minimum | Maximum | Mean | Median | Std. Dev. | Skewness | Kurtosis |
|---|---:|---:|---:|---:|---:|---:|---:|
| PM2.5 | 0.7 | 72.9 | 16.043 | 13.4 | 9.424 | 1.406 | 2.341 |
| PM10 | 1.1 | 74.3 | 18.056 | 15.9 | 9.472 | 1.268 | 2.072 |
| CO | 83 | 1935 | 262.592 | 225 | 150.805 | 3.840 | 23.034 |
| NO2 | 0 | 30.1 | 5.450 | 3.6 | 4.886 | 1.184 | 0.977 |
| SO2 | 0 | 45.5 | 7.878 | 5 | 7.356 | 1.270 | 1.285 |
| O3 | 15 | 254 | 89.832 | 84 | 32.187 | 1.240 | 2.196 |

### Kondisi Dataset

| Property | Nilai |
|---|---:|
| Row Count | 8.688 |
| Missing Values | 0 |
| NaN | 0 |
| +∞ | 0 |
| -∞ | 0 |

---

# 6. Analisis Hasil

### PM2.5

PM2.5 memiliki mean **16.043**, median **13.4**, dan skewness **1.406**. Hal ini menunjukkan distribusi PM2.5 cenderung miring ke kanan.

### PM10

PM10 memiliki mean **18.056**, median **15.9**, dan skewness **1.268**. Distribusi PM10 juga cenderung miring ke kanan.

### Carbon Monoxide (CO)

CO merupakan variabel yang paling menonjol.

- Mean = **262.592**
- Median = **225**
- Maximum = **1935**
- Skewness = **3.840**
- Kurtosis = **23.034**

Nilai skewness yang tinggi menunjukkan distribusi CO sangat miring ke kanan. Nilai kurtosis yang tinggi menunjukkan kecenderungan nilai ekstrem yang lebih kuat.

### Nitrogen Dioxide (NO2)

NO2 memiliki mean **5.450**, median **3.6**, dan skewness **1.184**. Distribusinya cenderung miring ke kanan.

### Sulphur Dioxide (SO2)

SO2 memiliki mean **7.878**, median **5**, dan skewness **1.270**. Distribusinya cenderung miring ke kanan.

### Ozone (O3)

O3 memiliki mean **89.832**, median **84**, dan skewness **1.240**. Distribusinya juga cenderung miring ke kanan.

---

# 7. Kesimpulan

Berdasarkan analisis menggunakan KNIME, dataset kualitas udara memiliki **8.688 observasi**.

Tidak terdapat:

- Missing values
- NaN
- Positive infinity
- Negative infinity

Sebagian besar variabel memiliki skewness positif sehingga distribusinya cenderung miring ke kanan.

Variabel yang paling menonjol adalah **Carbon Monoxide (CO)** dengan Skewness **3.840** dan Kurtosis **23.034**. Hal ini menunjukkan bahwa distribusi CO sangat miring ke kanan dan memiliki kecenderungan nilai ekstrem yang lebih kuat dibandingkan variabel lainnya.
