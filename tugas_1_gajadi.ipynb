```python
from pathlib import Path
import sys
import types


class LocalFiles:
    @staticmethod
    def upload():
        candidates = [
            Path("udara_klampis.geojson"),
            Path("materi") / "udara_klampis.geojson"
        ]

        for path in candidates:
            if path.exists():
                return {path.name: path.read_bytes()}

        raise FileNotFoundError(
            "File udara_klampis.geojson tidak ditemukan."
        )


# Menyediakan kompatibilitas google.colab.files saat dijalankan di VS Code.
google_module = types.ModuleType("google")
colab_module = types.ModuleType("google.colab")
colab_module.files = LocalFiles
google_module.colab = colab_module

sys.modules["google"] = google_module
sys.modules["google.colab"] = colab_module
```

```python
# ============================================================
# PROJECT SAINS DATA
# KUALITAS UDARA WILAYAH KLAMPIS
# KABUPATEN BANGKALAN
# ============================================================


# ============================================================
# 1. INSTALL LIBRARY
# ============================================================

import subprocess
import sys

def install_if_missing(packages):
    for package in packages:
        try:
            __import__(package)
        except ImportError:
            subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", package])

install_if_missing(["pandas", "requests", "folium", "matplotlib"])


# ============================================================
# 2. IMPORT LIBRARY
# ============================================================

import json
import requests
import pandas as pd
import folium
import matplotlib.pyplot as plt

from google.colab import files
from IPython.display import display


# ============================================================
# 3. KONFIGURASI POLUTAN
# ============================================================

# Semua polutan yang disimpan ke dalam dataset
kolom_polutan = [
    "pm2_5",
    "pm10",
    "carbon_monoxide",
    "nitrogen_dioxide",
    "sulphur_dioxide",
    "ozone"
]


# Polutan yang digunakan untuk grafik
# PM2.5 dan PM10 TIDAK dibuat grafik
polutan_grafik = [
    "carbon_monoxide",
    "nitrogen_dioxide",
    "sulphur_dioxide",
    "ozone"
]


# Nama yang ditampilkan pada grafik
nama_polutan = {
    "carbon_monoxide": "CO",
    "nitrogen_dioxide": "NO₂",
    "sulphur_dioxide": "SO₂",
    "ozone": "O₃"
}


# ============================================================
# 4. UPLOAD GEOJSON
# ============================================================

print("============================================")
print("UPLOAD GEOJSON WILAYAH KLAMPIS")
print("============================================")

uploaded = files.upload()

if not uploaded:
    raise ValueError(
        "Tidak ada file GeoJSON yang di-upload."
    )

nama_geojson = next(iter(uploaded))

print(
    "File yang digunakan:",
    nama_geojson
)


# ============================================================
# 5. MEMBACA GEOJSON
# ============================================================

with open(
    nama_geojson,
    "r",
    encoding="utf-8"
) as file:

    geojson = json.load(file)


# Memastikan GeoJSON memiliki feature
if (
    "features" not in geojson
    or len(geojson["features"]) == 0
):

    raise ValueError(
        "GeoJSON tidak memiliki feature."
    )


# Mengambil geometry
geometry = geojson["features"][0]["geometry"]

print(
    "Tipe geometri:",
    geometry["type"]
)


# Memastikan geometry berupa Polygon
if geometry["type"] != "Polygon":

    raise ValueError(
        "GeoJSON harus menggunakan Polygon."
    )


# ============================================================
# 6. MENGAMBIL KOORDINAT POLYGON
# ============================================================

coordinates = geometry["coordinates"][0]


# Menghapus titik terakhir jika merupakan
# pengulangan titik pertama
if coordinates[0] == coordinates[-1]:

    coordinates = coordinates[:-1]


# Format GeoJSON:
# [longitude, latitude]

longitudes = [
    point[0]
    for point in coordinates
]

latitudes = [
    point[1]
    for point in coordinates
]


# ============================================================
# 7. MENENTUKAN TITIK REPRESENTATIF WILAYAH
# ============================================================

longitude = (
    sum(longitudes)
    / len(longitudes)
)

latitude = (
    sum(latitudes)
    / len(latitudes)
)


print("\n============================================")
print("KOORDINAT WILAYAH")
print("============================================")

print(
    f"Latitude  : {latitude:.6f}"
)

print(
    f"Longitude : {longitude:.6f}"
)


# ============================================================
# 8. MEMBUAT PETA WILAYAH KLAMPIS
# ============================================================

peta = folium.Map(
    location=[
        latitude,
        longitude
    ],
    zoom_start=13
)


# Menampilkan batas GeoJSON
folium.GeoJson(
    geojson,
    name="Batas Wilayah Klampis"
).add_to(peta)


# Menambahkan titik representatif
folium.Marker(
    location=[
        latitude,
        longitude
    ],
    popup=(
        "Wilayah Klampis, "
        "Kabupaten Bangkalan"
    ),
    tooltip="Klampis"
).add_to(peta)


print("\n============================================")
print("PETA WILAYAH KLAMPIS")
print("============================================")

display(peta)


# ============================================================
# 9. OPEN-METEO AIR QUALITY API
# ============================================================

url = (
    "https://air-quality-api.open-meteo.com/v1/air-quality"
)


params = {

    "latitude": latitude,

    "longitude": longitude,

    "hourly": (
        "pm2_5,"
        "pm10,"
        "carbon_monoxide,"
        "nitrogen_dioxide,"
        "sulphur_dioxide,"
        "ozone"
    ),

    # Awal periode data
    "start_date": "2025-09-01",

    # Akhir periode data
    "end_date": "2026-08-28",

    "timezone": "Asia/Jakarta"
}


# ============================================================
# 10. MENGAMBIL DATA
# ============================================================

print("\n============================================")
print("MENGAMBIL DATA KUALITAS UDARA")
print("============================================")


response = requests.get(
    url,
    params=params,
    timeout=60
)


print(
    "Status API:",
    response.status_code
)


# Memastikan request berhasil
response.raise_for_status()


# Mengubah response menjadi JSON
data = response.json()


# ============================================================
# 11. MEMBUAT DATAFRAME
# ============================================================

df = pd.DataFrame(
    data["hourly"]
)


# Mengubah kolom waktu
df["time"] = pd.to_datetime(
    df["time"]
)


# ============================================================
# 12. MENAMBAHKAN INFORMASI WILAYAH
# ============================================================

df["wilayah"] = (
    "Wilayah Klampis, "
    "Kabupaten Bangkalan"
)

df["latitude"] = latitude

df["longitude"] = longitude


# ============================================================
# 13. MENYUSUN KOLOM DATASET
# ============================================================

df = df[
    [
        "time",
        "wilayah",
        "latitude",
        "longitude",
        "pm2_5",
        "pm10",
        "carbon_monoxide",
        "nitrogen_dioxide",
        "sulphur_dioxide",
        "ozone"
    ]
]


# ============================================================
# 14. MENGURUTKAN DATA
# ============================================================

df = (
    df
    .sort_values("time")
    .reset_index(drop=True)
)


# ============================================================
# 15. INFORMASI DATASET
# ============================================================

print("\n============================================")
print("DATA KUALITAS UDARA BERHASIL DIAMBIL")
print("============================================")

print(
    "Wilayah      :",
    df["wilayah"].iloc[0]
)

print(
    f"Latitude     : {latitude:.6f}"
)

print(
    f"Longitude    : {longitude:.6f}"
)

print(
    "Jumlah baris :",
    len(df)
)

print(
    "Jumlah kolom :",
    len(df.columns)
)

print(
    "Data awal    :",
    df["time"].min()
)

print(
    "Data akhir   :",
    df["time"].max()
)


# ============================================================
# 16. MENAMPILKAN DATA
# ============================================================

print("\n============================================")
print("10 DATA PERTAMA")
print("============================================")

display(
    df.head(10)
)


# ============================================================
# 17. MENYIMPAN DATA MENTAH KE CSV
# ============================================================

nama_csv = (
    "kualitas_udara_klampis.csv"
)


df.to_csv(
    nama_csv,
    index=False
)


print("\n============================================")
print("DATA MENTAH")
print("============================================")

print(
    "CSV berhasil dibuat:",
    nama_csv
)


# ============================================================
# 18. MEMBUAT DATA RATA-RATA HARIAN
# ============================================================

df_harian = df.copy()


# Membuat kolom tanggal
df_harian["tanggal"] = (
    df_harian["time"].dt.date
)


# Menghitung rata-rata setiap hari
df_harian = (
    df_harian
    .groupby("tanggal")[kolom_polutan]
    .mean()
    .reset_index()
)


# Mengubah tanggal menjadi datetime
df_harian["tanggal"] = pd.to_datetime(
    df_harian["tanggal"]
)


# ============================================================
# 19. MENAMPILKAN DATA RATA-RATA HARIAN
# ============================================================

print("\n============================================")
print("DATA RATA-RATA HARIAN")
print("============================================")

display(
    df_harian.head(10)
)


# ============================================================
# 20. MENYIMPAN DATA VISUALISASI KE CSV
# ============================================================

nama_csv_visualisasi = (
    "data_visualisasi_klampis.csv"
)


df_harian.to_csv(
    nama_csv_visualisasi,
    index=False
)


print("\n============================================")
print("DATA VISUALISASI")
print("============================================")




# ============================================================
# 21. GRAFIK TIME SERIES + PENJELASAN
# ============================================================

from IPython.display import display, Markdown

print("\n============================================")
print("GRAFIK TIME SERIES")
print("============================================")


# Penjelasan masing-masing grafik
penjelasan_grafik = {

    "carbon_monoxide": """
### Penjelasan Grafik CO

Konsentrasi CO cukup fluktuatif. Nilai relatif tinggi terlihat
pada akhir 2025 dan sekitar Maret 2026, dengan beberapa
lonjakan hingga sekitar 600 µg/m³.
""",

    "nitrogen_dioxide": """
### Penjelasan Grafik NO₂

NO₂ mengalami fluktuasi, dengan nilai relatif tinggi pada
akhir 2025. Memasuki pertengahan 2026, konsentrasinya
cenderung lebih rendah.
""",

    "sulphur_dioxide": """
### Penjelasan Grafik SO₂

SO₂ relatif rendah pada awal periode, kemudian meningkat
pada akhir 2025 hingga awal 2026. Setelah itu, konsentrasinya
cenderung menurun.
""",

    "ozone": """
### Penjelasan Grafik O₃

O₃ mengalami fluktuasi sepanjang periode. Beberapa nilai
tinggi terlihat pada akhir 2025 dan sekitar Maret 2026,
kemudian cenderung lebih stabil pada pertengahan 2026.
"""
}


# Membuat grafik satu per satu
for kolom in polutan_grafik:

    # --------------------------------------------------------
    # Membuat grafik
    # --------------------------------------------------------

    plt.figure(
        figsize=(14, 5)
    )

    plt.plot(
        df_harian["tanggal"],
        df_harian[kolom],
        linewidth=1
    )

    plt.xlabel(
        "Tanggal"
    )

    plt.ylabel(
        "Konsentrasi (µg/m³)"
    )

    plt.title(
        f"Time Series {nama_polutan[kolom]}\n"
        "Wilayah Klampis, Kabupaten Bangkalan"
    )

    plt.grid(
        True
    )

    plt.tight_layout()

    plt.show()


    # --------------------------------------------------------
    # Penjelasan di bawah grafik
    # --------------------------------------------------------

    display(
        Markdown(
            penjelasan_grafik[kolom]
        )
    )
```

### Penjelasan Setiap Polutan

| Polutan | Nama Lengkap | Penjelasan | Sumber Utama |
|---|---|---|---|
| PM₂.₅ | Fine Particulate Matter | Partikel sangat kecil dengan diameter ≤ 2,5 µm yang dapat masuk jauh ke dalam saluran pernapasan. | Kendaraan, pembakaran biomassa, industri, dan pembakaran bahan bakar |
| PM₁₀ | Particulate Matter ≤ 10 µm | Partikel dengan diameter ≤ 10 µm yang dapat masuk ke saluran pernapasan. | Debu jalan, konstruksi, kendaraan, industri, dan pembakaran |
| CO | Carbon Monoxide | Gas tidak berwarna dan tidak berbau yang terutama dihasilkan dari pembakaran bahan bakar yang tidak sempurna. | Kendaraan bermotor, mesin, dan proses pembakaran |
| NO₂ | Nitrogen Dioxide | Gas pencemar yang termasuk dalam kelompok nitrogen oksida (NOₓ) dan terutama dihasilkan dari proses pembakaran. | Kendaraan bermotor, pembangkit listrik, industri, dan pembakaran |
| SO₂ | Sulfur Dioxide | Gas pencemar yang terutama terbentuk ketika bahan bakar yang mengandung sulfur dibakar. | Pembangkit listrik, industri, batu bara, dan minyak |
| O₃ | Ozone | Ozon permukaan merupakan polutan sekunder yang terbentuk melalui reaksi kimia antara NOₓ dan senyawa organik volatil (VOC) dengan bantuan sinar matahari. | Emisi kendaraan dan industri yang bereaksi di atmosfer dengan bantuan sinar matahari |
